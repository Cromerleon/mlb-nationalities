```python
import statsapi as stp
import time as tm
from datetime import date as dt

actual_year = dt.today().year


anios = range(1956,actual_year + 1)
dominicans = {}

for anio in anios:
    try:
        data = stp.get('sports_players' , {'season': anio , 'gameType' : 'R'})
        for person in data.get('people' , []):
            country = person.get('birthCountry')
            if country == 'Dominican Republic':
                dominicans[person['id']] = person.get('fullName')
        
        print(f"{anio} : ok, total acumulado = {len(dominicans)}" )
    except Exception as e:
        print(f"{anio} : error -> {e}")
    
    tm.sleep(0.5)

print(f"\nTotal de jugadores dominicanos encontrados: {len(dominicans)}")
```

```python
import sqlite3 as sql

connection  = sql.connect('mlb_nacionalidades.db')

cursor = connection.cursor()

cursor.execute('''
               
    CREATE TABLE IF NOT EXISTS estadisticas_equipo(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    player_id INTEGER NOT NULL,
    team_id INTEGER NOT NULL,
    season INTEGER NOT NULL,
    tipo TEXT NOT NULL,           -- 'bateo' o 'pitcheo'
    AB INTEGER, H INTEGER, HR INTEGER, RBI INTEGER, AVG REAL, OBP REAL, SLG REAL,
    IP REAL, ERA REAL, W INTEGER, L INTEGER, SO INTEGER, WHIP REAL,
    FOREIGN KEY (player_id) REFERENCES jugadores(player_id),
    UNIQUE(player_id, team_id, season, tipo)
    )
               ''')
connection.commit()
connection.close()
```

```python
from datetime import date as dt

actual_year = dt.today().year

connection  = sql.connect('mlb_nacionalidades.db')

cursor = connection.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS jugadores_dominicanos (
    player_id INTEGER PRIMARY KEY,
    nombre_completo TEXT NOT NULL,
    fecha_nacimiento TEXT,
    ciudad_natal TEXT,
    posicion TEXT,
    fecha_debut TEXT,
    activo INTEGER,
    primer_anio_visto INTEGER,
    ultimo_anio_visto INTEGER,
    fecha_insercion TEXT DEFAULT CURRENT_TIMESTAMP
)
''')

years = range(1956 , actual_year + 1)


for year in years:
    try:
        data = stp.get('sports_players' , {'season': year , 'gameType' : 'R'})
        for p in data.get('people' , []):
            if p.get('birthCountry') != 'Dominican Republic':
                continue
            player_id = p['id']
            position = p.get('primaryPosition' , {}).get('name')
            
            cursor.execute('''
                    INSERT INTO jugadores_dominicanos
                        (player_id, nombre_completo, fecha_nacimiento, ciudad_natal,
                        posicion, fecha_debut, activo, primer_anio_visto, ultimo_anio_visto)
                    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
                    ON CONFLICT(player_id) DO UPDATE SET
                        ultimo_anio_visto = excluded.ultimo_anio_visto,
                        activo = excluded.activo
                ''', (
                    player_id, p.get('fullName'), p.get('birthDate'), p.get('birthCity'),
                    position, p.get('mlbDebutDate'), int(p.get('active', False)),
                    year, year
                ))
        
        connection.commit()
        print(f"{year}: ok")
    except Exception as e:
        
        print(f"{year} : error -> {e}")
    tm.sleep(0.5)
connection.close()
print("Ready")

```

```python
connection = sql.connect('mlb_nacionalidades.db')
cursor = connection.cursor()

def a_float(valor):
    try:
        return float(valor)
    except(TypeError, ValueError):
        return None

cursor.execute('select player_id from jugadores_dominicanos')
player_ids = [row[0] for row in cursor.fetchall()]
print(f"Total de jugaores a procesar: {len(player_ids)}")


for idx , pid in enumerate(player_ids):
    try:
        data = stp.get('people', {
            'personIds': pid,
            'hydrate': 'stats(group=[hitting,pitching],type=[yearByYear])'
        })
        person = data.get('people' , [])
        if not person:
            continue
        person = person[0]

        for stat_group in person.get('stats', []):
            group = stat_group.get('group' , {}).get('displayName')
            type = 'bateo' if group == 'hitting' else 'pitcheo' if group == 'pitching' else None
            if type is None:
                continue
            
            for split in stat_group.get('splits' , []):
                if 'team' not in split:
                    continue
                
                s = split['stat']
                team = split['team']
                season = int(split['season'])
                
                if type == 'bateo':
                        cursor.execute('''
                            INSERT INTO estadisticas_equipo
                                (player_id, team_id, team_name, season, tipo, AB, H, HR, RBI, AVG, OBP, SLG)
                            VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
                            ON CONFLICT(player_id, team_id, season, tipo) DO NOTHING
                        ''', (
                            pid, team['id'], team['name'], season, type,
                            s.get('atBats'), s.get('hits'), s.get('homeRuns'), s.get('rbi'),
                            a_float(s.get('avg')), a_float(s.get('obp')), a_float(s.get('slg'))
                        ))
                else:
                    cursor.execute('''
                        INSERT INTO estadisticas_equipo
                            (player_id, team_id, team_name, season, tipo, IP, ERA, W, L, SO, WHIP)
                        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
                        ON CONFLICT(player_id, team_id, season, tipo) DO NOTHING
                    ''', (
                        pid, team['id'], team['name'], season, type,
                        s.get('inningsPitched'), a_float(s.get('era')),
                        s.get('wins'), s.get('losses'), s.get('strikeOuts'), a_float(s.get('whip'))
                    ))    
        connection.commit()
        if idx % 50 == 0:
            print(f"{idx}/{len(player_ids)} procesados...")
    except Exception as e:
        print(f"player_id {pid} : error -> {e}")
    tm.sleep(0.5)

connection.close()
print("Llenado completado")
            
            
            
```
