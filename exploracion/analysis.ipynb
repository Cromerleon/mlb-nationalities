# Análisis: Jugadores de MLB por Nacionalidad

Este notebook analiza la distribución de jugadores activos de MLB por país de nacimiento, 
usando datos extraídos diariamente de la MLB Stats API.

Los datos se actualizan automáticamente todos los días vía GitHub Actions.

```python
import pandas as pd
import sqlite3 as sql
from datetime import date as dt

today = dt.today().isoformat()

connection_bd = sql.connect('mlb_nacionalidades.db')

df_players = pd.read_sql_query('select * from jugadores' , connection_bd)
df_history = pd.read_sql_query('select * from historial_diario',connection_bd)

connection_bd.close()

print(f"Total de jugadores: {len(df_players)}")
print(f"Última actualización: {df_players['fecha_actualizacion'].max()}")
```

## Distribución de jugadores por país

```python
count_by_country = df_players['pais_nacimiento'].value_counts()
print(count_by_country)
```

## Distribución de Jugadores por Ciudad

```python
count_by_city = df_players['ciudad_nacimiento'].value_counts()
print(count_by_city)
```

## Jugadores por equipo y nacionalidad

```python
cross_table = pd.crosstab(df_players['equipo_nombre'] , df_players['pais_nacimiento'])
cross_table
```
